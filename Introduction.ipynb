## Who is this book for?

This book is written for undergraduate students who want to learn the basics of robotics and get hands-on experience with real robots. It is also a good fit for anyone else who wants to learn the fundamentals of robotics from the ground up. You will need some calculus, linear algebra, and basic programming. The book introduces the Python you need along the way.

It is **not** aimed at experienced researchers. If you are already comfortable with robotics fundamentals and want a more advanced or comprehensive treatment, we recommend *Modern Robotics: Mechanics, Planning, and Control* (Lynch and Park) or *Robot Modeling and Control* (Spong, Hutchinson, and Vidyasagar).
